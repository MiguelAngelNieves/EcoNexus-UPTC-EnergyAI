```python
import warnings

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# Evita que salgan advertencias
warnings.filterwarnings("ignore")

# Muestra más información cuando se imprime un DataFrame
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

# Estilo simple para gráficos
plt.style.use("seaborn-v0_8-darkgrid")
sns.set_palette("husl")
```

```python
# Cargar archivos CSV
ruta_consumos = "../data/raw/consumos_uptc.csv"
ruta_sedes = "../data/raw/sedes_uptc.csv"

print("Cargando archivo de consumos...")
df_consumos = pd.read_csv(ruta_consumos)
print("Archivo de consumos cargado.")

filas, columnas = df_consumos.shape
print(f"Filas: {filas} | Columnas: {columnas}")

print("Cargando archivo de sedes...")
df_sedes = pd.read_csv(ruta_sedes)
print("Archivo de sedes cargado.")

cantidad_sedes = len(df_sedes)
print(f"Sedes cargadas: {cantidad_sedes}")
```

```python
# Revisar primeras y últimas filas del dataset
print("Primeras 10 filas del dataset de consumos:")
display(df_consumos.head(10))

print("Últimas 10 filas del dataset de consumos:")
display(df_consumos.tail(10))
```

```python
# Información general del dataset
print("Información general del dataset de consumos:")
df_consumos.info()

print("Cantidad de columnas por tipo de dato:")
tipos_datos = df_consumos.dtypes.value_counts()
print(tipos_datos)
```

```python
# Revisar columnas, tipos de datos y valores nulos

print("Listado de columnas del dataset de consumos:")

info_columnas = pd.DataFrame({
    "columna": df_consumos.columns,
    "tipo_dato": df_consumos.dtypes.values,
    "valores_no_nulos": df_consumos.count().values,
    "valores_nulos": df_consumos.isnull().sum().values,
    "porcentaje_nulos": (df_consumos.isnull().sum() / len(df_consumos) * 100).values
})

info_columnas["porcentaje_nulos"] = info_columnas["porcentaje_nulos"].round(2)

display(info_columnas)

total_columnas = len(df_consumos.columns)
columnas_con_nulos = (info_columnas["porcentaje_nulos"] > 0).sum()
columnas_con_muchos_nulos = (info_columnas["porcentaje_nulos"] > 20).sum()

print(f"Total de columnas: {total_columnas}")
print(f"Columnas con valores nulos: {columnas_con_nulos}")
print(f"Columnas con más del 20% de valores nulos: {columnas_con_muchos_nulos}")
```

```python
# Estadísticas descriptivas
print("Estadísticas de columnas numéricas:")
estadisticas_numericas = df_consumos.describe().T
display(estadisticas_numericas)

print("Estadísticas de columnas categóricas:")
estadisticas_categoricas = df_consumos.describe(include=["object"])
display(estadisticas_categoricas)
```

```python
# Analisis temporal usando la columna timestamp
df_consumos["timestamp"] = pd.to_datetime(df_consumos["timestamp"])

print("Analisis temporal del dataset de consumos")

fecha_inicio = df_consumos["timestamp"].min()
fecha_fin = df_consumos["timestamp"].max()
rango_dias = (fecha_fin - fecha_inicio).days

print(f"Fecha inicial: {fecha_inicio}")
print(f"Fecha final: {fecha_fin}")
print(f"Rango de días: {rango_dias}")

anios_unicos = df_consumos["timestamp"].dt.year.nunique()
meses_unicos = df_consumos["timestamp"].dt.to_period("M").nunique()

print(f"Años diferentes en los datos: {anios_unicos}")
print(f"Meses diferentes en los datos: {meses_unicos}")

print("Cantidad de registros por año:")
registros_por_anio = df_consumos["timestamp"].dt.year.value_counts().sort_index()
print(registros_por_anio)
```

```python
# Analisis y visualización por sede

print("Distribución de registros por sede")

conteo_sedes = df_consumos["sede"].value_counts()
porcentaje_sedes = df_consumos["sede"].value_counts(normalize=True) * 100

distribucion_sedes = pd.DataFrame({
    "sede": conteo_sedes.index,
    "registros": conteo_sedes.values,
    "porcentaje": porcentaje_sedes.values
})

display(distribucion_sedes)

fig, ejes = plt.subplots(1, 2, figsize=(14, 5))

# Gráfico de barras
ejes[0].bar(distribucion_sedes["sede"], distribucion_sedes["registros"])
ejes[0].set_xlabel("Sede")
ejes[0].set_ylabel("Cantidad de registros")
ejes[0].set_title("Registros por sede")
ejes[0].tick_params(axis="x", rotation=45)

# Gráfico circular
ejes[1].pie(
    distribucion_sedes["porcentaje"],
    labels=distribucion_sedes["sede"],
    autopct="%1.1f%%",
    startangle=90
)
ejes[1].set_title("Distribución porcentual por sede")

plt.tight_layout()
plt.show()
```

```python
# Analisis de energía total
print("Analisis de energía total (kWh)")

serie_energia_total = df_consumos["energia_total_kwh"]

estadisticas_energia = serie_energia_total.describe()
print(estadisticas_energia)

valor_minimo = serie_energia_total.min()
valor_maximo = serie_energia_total.max()
rango_valores = valor_maximo - valor_minimo
coef_variacion = serie_energia_total.std() / serie_energia_total.mean()

print(f"Valor mínimo: {valor_minimo:.2f} kWh")
print(f"Valor máximo: {valor_maximo:.2f} kWh")
print(f"Rango: {rango_valores:.2f} kWh")
print(f"Coeficiente de variación: {coef_variacion:.2%}")

# Gráficas
fig, ejes = plt.subplots(2, 2, figsize=(14, 10))

# Histograma
valores_histograma = serie_energia_total.dropna()
ejes[0, 0].hist(valores_histograma, bins=50, edgecolor="black")
ejes[0, 0].set_xlabel("kWh")
ejes[0, 0].set_ylabel("Frecuencia")
ejes[0, 0].set_title("Distribución de energía total")
ejes[0, 0].axvline(serie_energia_total.mean(), linestyle="--", label="Media")
ejes[0, 0].axvline(serie_energia_total.median(), linestyle="--", label="Mediana")
ejes[0, 0].legend()

# Boxplot
ejes[0, 1].boxplot(valores_histograma)
ejes[0, 1].set_ylabel("kWh")
ejes[0, 1].set_title("Boxplot: energía total")

# Boxplot por sede
df_consumos.boxplot(column="energia_total_kwh", by="sede", ax=ejes[1, 0])
ejes[1, 0].set_xlabel("Sede")
ejes[1, 0].set_ylabel("kWh")
ejes[1, 0].set_title("Energía total por sede")
plt.sca(ejes[1, 0])
plt.xticks(rotation=45)

# Q-Q plot (revisión visual de normalidad)
from scipy import stats

stats.probplot(valores_histograma, dist="norm", plot=ejes[1, 1])
ejes[1, 1].set_title("Q-Q plot: energía total")

plt.tight_layout()
plt.show()
```

```python
# Analisis por sectores
print("Analisis de consumo por sectores")

columnas_sectores = [
    "energia_comedor_kwh",
    "energia_salones_kwh",
    "energia_laboratorios_kwh",
    "energia_auditorios_kwh",
    "energia_oficinas_kwh",
]

estadisticas_sectores = df_consumos[columnas_sectores].describe().T

nulos_por_sector = df_consumos[columnas_sectores].isnull().sum()
porcentaje_nulos_sector = (nulos_por_sector / len(df_consumos) * 100).round(2)

estadisticas_sectores["nulos"] = nulos_por_sector.values
estadisticas_sectores["porcentaje_nulos"] = porcentaje_nulos_sector.values

display(estadisticas_sectores)

print("Consumo promedio por sector:")
consumo_promedio_sector = df_consumos[columnas_sectores].mean().sort_values(ascending=False)
print(consumo_promedio_sector)

# Gráficas
fig, ejes = plt.subplots(1, 2, figsize=(14, 5))

# Barras horizontales
consumo_promedio_sector.plot(kind="barh", ax=ejes[0], edgecolor="black")
ejes[0].set_xlabel("Consumo promedio (kWh)")
ejes[0].set_title("Consumo promedio por sector")

# Boxplot comparativo
df_consumos[columnas_sectores].plot(kind="box", ax=ejes[1])
ejes[1].set_ylabel("kWh")
ejes[1].set_title("Distribución de consumo por sector")
ejes[1].tick_params(axis="x", rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Patrón de consumo por hora y por día de semana

print("Patrón de consumo por hora")

consumo_por_hora = df_consumos.groupby("hora")["energia_total_kwh"].agg(["mean", "median", "std"])
display(consumo_por_hora)

fig, ejes = plt.subplots(1, 2, figsize=(14, 5))

# Línea: consumo promedio y mediana por hora
ejes[0].plot(consumo_por_hora.index, consumo_por_hora["mean"], marker="o", linewidth=2, label="Promedio")
ejes[0].plot(consumo_por_hora.index, consumo_por_hora["median"], marker="s", linewidth=2, label="Mediana")

# Banda de variación usando la desviación estándar
ejes[0].fill_between(
    consumo_por_hora.index,
    consumo_por_hora["mean"] - consumo_por_hora["std"],
    consumo_por_hora["mean"] + consumo_por_hora["std"],
    alpha=0.3,
    label="± 1 desviación estándar",
)

ejes[0].set_xlabel("Hora del día")
ejes[0].set_ylabel("Consumo (kWh)")
ejes[0].set_title("Consumo por hora")
ejes[0].legend()
ejes[0].set_xticks(range(0, 24))

# Heatmap: consumo promedio por hora vs día de la semana
consumo_hora_dia = df_consumos.pivot_table(
    values="energia_total_kwh",
    index="hora",
    columns="dia_nombre",
    aggfunc="mean",
)

dias_ordenados = ["Lunes", "Martes", "Miércoles", "Jueves", "Viernes", "Sábado", "Domingo"]
consumo_hora_dia = consumo_hora_dia[dias_ordenados]

sns.heatmap(
    consumo_hora_dia,
    annot=False,
    fmt=".0f",
    ax=ejes[1],
    cbar_kws={"label": "kWh"},
)

ejes[1].set_xlabel("Día de la semana")
ejes[1].set_ylabel("Hora del día")
ejes[1].set_title("Consumo por hora y día")

plt.tight_layout()
plt.show()
```

```python
# Resumen de exploración inicial
print("Resumen de exploración inicial")

total_registros = len(df_consumos)
fecha_inicio = df_consumos["timestamp"].min()
fecha_fin = df_consumos["timestamp"].max()

sedes_unicas = df_consumos["sede"].nunique()
lista_sedes = ", ".join(df_consumos["sede"].dropna().unique().astype(str))

columnas_numericas = df_consumos.select_dtypes(include=[np.number]).shape[1]
columnas_categoricas = df_consumos.select_dtypes(include=["object"]).shape[1]

columnas_con_nulos = (df_consumos.isnull().sum() > 0).sum()

print(f"Total de registros: {total_registros:,}")
print(f"Periodo: {fecha_inicio} a {fecha_fin}")
print(f"Sedes: {sedes_unicas} ({lista_sedes})")
print(f"Columnas numéricas: {columnas_numericas}")
print(f"Columnas categóricas: {columnas_categoricas}")

print(f"- Columnas con valores nulos: {columnas_con_nulos}")
```
